## 2 数值与算术运算

Python 区分整数（`int`）与浮点数（`float`）。`type()` 返回一个值的类型。

```python
print(type(100))
print(type(100.0))
```

| 运算符 | 含义 | 示例 |
|---|---|---|
| `+` `-` `*` | 加、减、乘 | `3 * 4` |
| `/` | 除，结果总是浮点数 | `7 / 2` |
| `//` | 整除，向下取整 | `7 // 2` |
| `%` | 取余 | `7 % 2` |
| `**` | 乘方 | `2 ** 10` |

```python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
```

`/` 的结果总是浮点数，即使能整除。

```python
print(6 / 2)
```

浮点数是实数的近似表示，运算可能有微小误差。比较两个浮点数是否相等时不能直接用 `==`。

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

科学计数法：`7.9e-11` 表示 7.9 × 10⁻¹¹。

```python
print(7.9e-11)
print(0.79e-10)
```

## 3 变量与赋值

`名字 = 值` 把一个名字与一个值关联起来。`=` 不是数学上的等号，而是“先算右边，再把结果贴上左边的名字”这一动作。

```python
height = 1.79
weight = 68.7
bmi = weight / height ** 2
print(bmi)
```

名字可以重新赋值。下面把 `weight` 改成 70.0，再重新计算 `bmi`。注意：如果只改 `weight` 而不重跑第三行，`bmi` 不会自动更新。

```python
weight = 70.0
bmi = weight / height ** 2
print(bmi)
```

## 4 字符串

用引号括起来的文本是字符串（`str`）。单引号与双引号等价。`len()` 返回长度。

```python
name = "Zhejiang"
print(name)
print(type(name))
print(len(name))
```

`+` 与 `*` 对字符串也有定义，但含义不同：`+` 是拼接，`*` 是重复。同一个运算符对不同类型的值含义不同。

```python
print("Zhejiang" + " " + "University")
print("ab" * 3)
```

## 5 对象与方法

Python 中每个值都是对象，对象自带一组操作，称为方法。记法是 `对象.方法名()`。本课程后面绝大多数代码都是这种形式，例如 `nav.head()`、`nav.groupby("基金代码")`。

```python
print(bmi.is_integer())
print(name.upper())
print(name.startswith("Zhe"))
```

不同类型的对象有不同的方法。下面这一格**会报错**，是有意安排的：字符串没有 `is_integer()` 方法。

```python
name.is_integer()
```
