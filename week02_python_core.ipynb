## 1 值、类型与变量

### 1.1 值与类型

Python 程序处理的每一项数据称为一个值（value）。第一讲遇到过三类值：整数（如 `42`）、浮点数（如 `1.2345`）和字符串（如 `"000001"`）。每个值都属于一种类型（type），这三类分别是 `int`、`float` 和 `str`。内置函数 `type()` 返回一个值的类型。

```python
nav = 1.2345
code = "000001"
n = 42
print(type(nav))
print(type(code))
print(type(n))
```

类型之所以重要，是因为它决定了一个值可以参与哪些运算，以及运算的含义。第一讲提到过，字符串有 `upper()`、`zfill()` 等方法，而数字没有；反过来，`round()` 可以作用于浮点数，却不能作用于字符串。

```python
print(round(nav, 2))
print(code.zfill(8))
print(len(code))
```

同一个运算符作用于不同类型，含义可能不同。加号 `+` 对两个数字是加法，对两个字符串是拼接；乘号 `*` 对两个数字是乘法，对字符串和整数则是重复。

```python
print(1 + 2)
print("1" + "2")
print("ab" * 3)
```

如果两个操作数的类型不匹配，Python 不会猜测你的意图，而是报错。下面这一格会报错。

```python
"1" + 2
```

错误信息的最后一行是 `TypeError: can only concatenate str (not "int") to str`。`TypeError` 是错误的类别，表示类型不匹配；冒号后面的文字说明具体原因：字符串只能与字符串拼接，不能与整数拼接。读错误信息时，先看最后一行的错误类别，再看它后面的说明，最后看上方箭头指向的是哪一行代码。

### 1.2 变量与赋值

语句 `nav = 1.2345` 称为赋值（assignment）。它的执行分两步：先计算等号右边的表达式，得到一个值；再把等号左边的名字与这个值关联起来。此后在代码中写 `nav`，Python 就用它所关联的值来代替。名字可以重新赋值，指向另一个值。

```python
x = 1
print(x)
x = x + 1
print(x)
```

第三行 `x = x + 1` 初学时容易困惑，因为它不是数学等式。按赋值的两步理解就清楚了：先计算右边 `x + 1`，此时 `x` 是 1，结果是 2；再把名字 `x` 关联到 2。

变量名由字母、数字和下划线组成，不能以数字开头，区分大小写；`nav`、`Nav`、`NAV` 是三个不同的名字。`if`、`for`、`def` 等 Python 保留字不能作变量名。此外，尽量不要用 `list`、`sum`、`type` 这类内置函数的名字作变量名，否则会把原来的函数覆盖掉。

### 1.3 布尔值与比较运算

除了数字和字符串，还有一种只有两个取值的类型：布尔类型 `bool`，取值为 `True` 或 `False`。比较运算的结果就是布尔值。比较运算符有六个：`>`、`<`、`>=`、`<=`、`==`（等于）和 `!=`（不等于）。注意“等于”是两个等号，单个等号是赋值。

```python
print(3 > 2)
print(3 == 2)
print(3 != 2)
print(type(3 > 2))
```

多个条件可以用 `and`、`or`、`not` 组合。`a and b` 当且仅当两者都为真时为真；`a or b` 只要有一个为真就为真；`not a` 取反。判断一个数是否落在某个区间，可以写成 `r > 0 and r < 0.05`，也可以写成数学上更自然的 `0 < r < 0.05`。

```python
r = 0.02
print(r > 0 and r < 0.05)
print(r > 0 or r < -0.05)
print(not r > 0)
print(0 < r < 0.05)
```

布尔值参与算术运算时，`True` 按 1 计算，`False` 按 0 计算。因此对一组布尔值求和，得到的是其中 `True` 的个数。这个性质在第 8 节统计“有多少个元素满足条件”时会用到。

```python
print(True + True)
print(sum([True, False, True]))
```

布尔值是第 6 节条件语句和第 8 节数据筛选的基础。

### 1.4 空值 None

`None` 是一个特殊的值，表示“没有值”。它属于类型 `NoneType`。判断一个变量是否为 `None`，习惯上写 `is None` 而不是 `== None`。第 7 节会看到它最常出现的场合：一个函数没有返回任何结果时，调用它得到的就是 `None`。

```python
result = None
print(result)
print(type(result))
print(result is None)
```

### 1.5 类型转换

函数 `int()`、`float()`、`str()` 把一个值转换为相应的类型。数据从文件读入时往往是字符串，需要转换成数字才能计算；反过来，把数字拼进一句话里时，需要先转成字符串。

```python
print(int("42"))
print(float("3.5"))
print(str(42) + " 只基金")
print(int(3.99))       # 向零截断，不是四舍五入
```

有一类数据需要特别注意：基金代码、股票代码、邮政编码、身份证号等。它们由数字字符组成，但并不是数量，只是用来识别对象的编码。编码不参与算术，应当作为字符串保存。一旦被转换成整数，开头的零就会丢失，而且无法自动恢复。

```python
code = "000003"
n = int(code)
print(n)
print(str(n))
print(str(n).zfill(6))    # 只有事先知道编码是 6 位，才能补回来
print(code == 3)
print(code == "000003")
```

`code == 3` 的结果是 `False`：字符串 `"000003"` 与整数 `3` 是两个不同类型的值，即使看起来相近，也不相等。第 3 周读入数据表时，如果基金代码一列显示为 `3`、`150003` 这样的整数，就是读入时被当成了数字，需要指定为字符串重新读入。

### 1.6 浮点数与算术运算

先看一个出乎意料的结果。

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

0.1 加 0.2 得到的不是 0.3，而是 0.30000000000000004。要理解这个结果，需要知道计算机是怎样保存小数的。

我们熟悉的十进制中，小数点后的各位分别表示十分之一、百分之一、千分之一……有些分数用十进制小数写不完：1/3 写成 0.3333…，无论写多少位都只是近似值，写到某一位截断就会产生误差。计算机内部使用二进制，小数点后的各位分别表示 1/2、1/4、1/8、1/16……在这个系统中，能够精确写出的只有那些可以拆成有限个 1/2、1/4、1/8 …之和的数：0.5 就是 1/2，0.75 是 1/2 + 1/4，0.625 是 1/2 + 1/8。而 0.1 即 1/10，无法拆成有限个这样的分数之和，它的二进制表示是无限循环的 0.0001100110011…，就像 1/3 在十进制中无限循环一样。

计算机给每个浮点数分配的存储空间是固定的，只能保留约 53 位二进制有效数字（相当于十进制的 15 到 16 位），超出的部分被舍去。因此 0.1 存入计算机后，实际保存的是一个非常接近、但不等于 0.1 的数。用格式化输出保留 20 位小数，可以看到它们真正保存的值。

```python
# f"..." 是格式化字符串：花括号里 0.1 是要输出的值，冒号后的 .20f 表示保留 20 位小数
print(f"{0.1:.20f}")
print(f"{0.2:.20f}")
print(f"{0.3:.20f}")
```

三个数各自带有约在第 17 位小数上的误差。0.1 与 0.2 的误差相加，与 0.3 自身的误差恰好不同，于是 `0.1 + 0.2 == 0.3` 为假。同一原因，`0.1 * 3` 也不等于 0.3。相比之下，能够拆成 1/2、1/4、1/8 …之和的数在二进制中是精确的，它们之间的运算不会出现这种情况。

```python
print(0.1 * 3 == 0.3)
print(0.5 + 0.25 == 0.75)
```

这种误差出现在第 17 位小数左右，对金融计算的数值本身几乎没有影响；它真正造成麻烦的地方是相等判断。两个本应相等的计算结果，因为中间步骤的舍入路径不同，可能在最后一位上差一点，用 `==` 比较就得到 `False`。因此判断两个浮点数是否“相等”，不应直接用 `==`，而应检查两者之差的绝对值是否小于一个很小的容差。

```python
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
```

以后若遇到“两列收益率按理应当相同，逐个比较却有几行为 `False`”，多半就是这个原因。

其余算术运算符：`/` 是除法，结果总是浮点数；`//` 是整除，舍去小数部分；`%` 是取余；`**` 是乘方。

```python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
```

到此为止，每个变量只保存一个值。但数据分析处理的是成批的数据：一只基金几年的日净值有上千个数，不可能为每个数单独命名。我们需要一种能把许多值放在一起、用一个名字整体引用的对象。这类对象统称容器（container）。Python 内置了四种容器：列表、元组、字典和集合。下面先介绍最常用的列表。

## 2 列表

### 2.1 创建列表

列表（list）是一组按顺序排列的值，写在方括号内，元素之间用逗号分隔。下面用一个列表保存某只基金连续四期的净值。

```python
navs = [1.00, 1.05, 1.02, 1.10]
print(navs)
print(type(navs))
print(len(navs))
```

`len()` 返回列表的长度，即元素个数。空列表写作 `[]`。

列表对元素的类型没有限制：数字、字符串、布尔值都可以放，同一个列表里的元素类型也可以各不相同。

```python
mixed = [1, "a", 2.5, True]
empty = []
print(len(mixed), len(empty))
print(type(mixed[0]), type(mixed[1]))
```

不过在数据处理中，一列数据通常应当是同一类型；一列净值里混进一个字符串（例如用 `"--"` 表示缺失），往往是数据出了问题的信号，而且会让后面的算术运算报错。

列表的元素还可以是另一个列表，形成嵌套结构。一张数据表就可以这样表示：外层列表的每个元素是一行，每一行本身又是一个列表。

```python
table = [["000001", "华夏成长", 1.2345],
         ["000003", "中海可转债", 0.9876]]
print(len(table))          # 2 行
print(table[0])            # 第 0 行，是一个列表
print(table[0][2])         # 第 0 行中索引为 2 的字段
```

`table[0][2]` 是连续两次索引：先取出第 0 行（一个列表），再从这个列表中取索引为 2 的元素。第 8 节的二维数组和第 3 周的数据表，本质上都是这种“行的序列”的更高效的版本。

### 2.2 索引：按位置取出元素

列表中的每个元素都有一个位置编号，称为索引（index）。索引从 0 开始：第一个元素的索引是 0，第二个是 1，依此类推。取出某个位置的元素，在列表名后写方括号和索引。对于 `navs = [1.00, 1.05, 1.02, 1.10]`：

| 元素 | 1.00 | 1.05 | 1.02 | 1.10 |
| --- | --- | --- | --- | --- |
| 索引 | 0 | 1 | 2 | 3 |
| 负索引 | −4 | −3 | −2 | −1 |

```python
print(navs[0])
print(navs[1])
print(navs[3])
```

索引也可以是负数，表示从末尾倒数：`-1` 是最后一个元素，`-2` 是倒数第二个。当不知道列表长度、又想取最后一个元素时，`navs[-1]` 比 `navs[len(navs) - 1]` 简洁。

```python
print(navs[-1])
print(navs[-2])
```

四个元素的索引是 0、1、2、3。访问不存在的位置会报错。下面这一格会报错。

```python
navs[4]
```

`IndexError: list index out of range`，意思是索引超出了范围。这是初学时最常见的错误之一，多数情况下是忘记了索引从 0 开始。

### 2.3 切片：取出一段

从列表中取出连续的一段，称为切片（slice），写法是 `[起点:终点]`。切片包含起点位置的元素，但不包含终点位置的元素，即“含头不含尾”。这样规定的好处是：切片的长度恰好等于终点减起点，而且 `navs[:k]` 与 `navs[k:]` 正好把列表分成前 k 个和其余的两部分，没有重叠也没有遗漏。

```python
print(navs[1:3])     # 索引 1、2 两个元素
print(navs[:2])      # 省略起点，从头开始：索引 0、1
print(navs[2:])      # 省略终点，直到末尾：索引 2、3
```

终点也可以是负数。`navs[:-1]` 表示从头到倒数第二个，即去掉最后一个元素；`navs[1:]` 表示去掉第一个元素。

```python
print(navs[:-1])
print(navs[1:])
```

请注意这两个切片。原列表有 4 个元素，它们各有 3 个。把它们按位置对齐，`navs[:-1]` 的每个元素是“前一期净值”，`navs[1:]` 对应位置的元素是“后一期净值”。第 6 节计算收益率时就利用这种对齐。

切片还可以带第三个参数——步长。`navs[::2]` 每隔一个元素取一个；`navs[::-1]` 步长为 −1，得到倒序的列表。

```python
print(navs[::2])
print(navs[::-1])
```

### 2.4 修改列表

列表建立后可以修改：给某个位置赋新值，或者增加、删除元素。第一讲说过，方法是属于某种类型的函数，通过“对象.方法名()”调用。列表常用的修改方法如下。

```python
navs[2] = 1.03            # 把索引 2 的元素改为 1.03
print(navs)
```

在末尾增加元素有两个方法，`append()` 和 `extend()`，初学时很容易混用。`append(x)` 只接受一个参数，把 x 作为一个元素加到末尾——不管 x 本身是什么。传入一个数，就加入一个数。

```python
navs.append(1.12)
print(navs)
print(len(navs))
```

如果传给 `append()` 的是一个列表，整个列表会作为一个元素放进去，结果是列表套列表，长度只增加 1。这通常不是想要的结果。

```python
navs.append([1.15, 1.11])
print(navs)
print(len(navs))
print(navs[-1])            # 最后一个元素本身是一个列表
```

`extend(x)` 则把参数 x 看作一组元素，先把它拆开，再把其中的元素逐个加到末尾。用它加入两个净值，长度增加 2，列表仍然是平的。

```python
navs.pop()                 # 先删掉刚才误加进去的子列表（pop 见下文）
navs.extend([1.15, 1.11])
print(navs)
print(len(navs))
```

两者的区别可以概括为：`append(x)` 加入的是 x 本身，一个；`extend(x)` 加入的是 x 里面的东西，若干个。`extend()` 的参数可以是任何能够逐个取出元素的对象——列表、元组，甚至字符串。传入字符串时，它会被拆成单个字符，这一点常常出人意料。

```python
letters = ["a", "b"]
letters.extend("cd")       # 字符串被拆成 "c"、"d" 两个元素
print(letters)
letters.append("ef")       # 字符串 "ef" 整体作为一个元素
print(letters)
```

加号 `+` 连接两个列表，效果与 `extend()` 类似，但它返回一个新列表，原列表不变；这一区别在 2.5 节讨论。

```python
a = [1, 2]
b = a + [3, 4]
print(a)
print(b)
```

其余几个修改方法：`insert()` 在指定位置插入，`pop()` 取出并删除末尾（或指定位置）的元素，`remove()` 删除第一个等于给定值的元素。

```python
navs.insert(0, 0.98)       # 在索引 0 处插入，原有元素后移
print(navs)
last = navs.pop()          # 取出并删除最后一个元素
print(last)
print(navs)
navs.remove(0.98)          # 删除第一个值为 0.98 的元素
print(navs)
```

### 2.5 返回新值与原地修改

第一讲说过，字符串的方法总是返回新值，因为字符串本身不能修改。列表可以修改，于是对列表的操作就分成两类：一类算出一个新值交出来，原列表不动；另一类直接改动原列表，不交出有用的值。分类的依据是“原列表有没有被改”，与写成函数还是方法无关——两类操作都既有函数也有方法。

以排序为例。内置函数 `sorted(p)` 接收列表 `p`，返回一个新的、排好序的列表，`p` 保持不变。

```python
p = [3, 1, 2]
s = sorted(p)
print(s)
print(p)
```

列表自己的方法 `p.sort()` 则直接把 `p` 本身排好序，称为原地（in place）修改。原地修改的操作不返回排好序的列表——它的返回值是 `None`。

```python
p = [3, 1, 2]
r = p.sort()
print(r)
print(p)
```

因此 `r = p.sort()` 几乎总是写错了：想得到一个新的有序列表应写 `sorted(p)`；想把 `p` 本身排好序只写 `p.sort()`，不接收返回值。

列表的方法本身也分属两类。上一小节的 `append`、`extend`、`insert`、`remove` 都是原地修改，返回 `None`；而 `count(x)`（x 出现的次数）、`index(x)`（x 第一次出现的位置）、`copy()`（复制一份）只是算出并返回一个值，不改动原列表。

```python
p = [3, 1, 2, 1]
print(p.count(1))
print(p.index(2))
q = p.copy()
print(q)
print(p)          # p 没有被上面三行改动
```

读代码遇到一个函数或方法时，需要分清它属于哪一类：返回一个新值而不改动原对象，还是改动原对象而不返回有用的值。不确定时，把返回值和原对象都打印出来看一看。

### 2.6 变量与对象：赋值并不复制

到这里需要更准确地说明变量与值之间的关系。初学者常把变量理解为一个“盒子”，赋值就是把值装进盒子；`b = a` 就是把 `a` 盒子里的东西复制一份放进 `b` 盒子。这个理解在下面的例子上会得出错误的预测。

```python
a = [1, 2]
b = a
b.append(3)
print(a)
```

`a` 也变成了 `[1, 2, 3]`，尽管代码没有直接改动 `a`。原因是：在 Python 中，值（更准确地说是对象）独立存在于内存中，变量只是指向对象的名字。`a = [1, 2]` 建立了一个列表对象并让 `a` 指向它；`b = a` 并不复制列表，只是让 `b` 也指向同一个对象。此后无论通过 `a` 还是 `b` 修改，改动的都是这唯一的一个列表。

这里的“指向”就是其他语言中所说的引用或指针：变量里存的不是数据本身，而是数据在内存中的位置。Python 不允许直接操作地址，但内置函数 `id()` 会返回每个对象的编号（在通常的 Python 实现中就是它的内存地址）。两个名字指向同一个对象，编号就相同。

```python
a = [1, 2]
b = a
print(id(a))
print(id(b))
print(id(a) == id(b))
```

但下面这段代码看起来与此矛盾。

```python
x = 5
y = x
y = y + 1
print(x)
print(y)
```

`x` 没有随 `y` 改变。这不是因为数字的行为不同，而是因为整数是不可变（immutable）对象：`y = y + 1` 不能修改对象 5 本身，只能新建一个对象 6，再让 `y` 指向它，`x` 仍然指向 5。列表则是可变（mutable）对象，`append` 直接改动对象本身，所以指向它的所有名字都看到了改动。

Python 的内置类型可以按此分类：整数、浮点数、布尔值、字符串和元组是不可变的；列表、字典、集合是可变的。以后要学的 NumPy 数组和 pandas 数据表也是可变的。对可变对象而言，`b = a` 之后通过 `b` 的修改会反映到 `a` 上；要得到一个独立的副本，必须显式复制，例如 `a.copy()` 或 `list(a)`。复制会在内存中新建一个列表对象，所以副本的编号与原列表不同。

```python
a = [1, 2]
b = a.copy()
print(id(a) == id(b))
b.append(3)
print(a)
print(b)
```

`copy()` 新建的只是外层的列表；列表里的各个元素没有被复制，副本中存放的仍是指向原来那些元素的引用。元素是数字、字符串这类不可变对象时这没有任何影响。但如果元素本身是可变对象——例如 2.1 节那种“列表的列表”，副本与原表就共享每一行；通过副本修改某一行，原表也会跟着变。这种只复制一层的复制称为浅复制。

```python
table = [["000001", 1.2345], ["000003", 0.9876]]
table2 = table.copy()
table2[0][1] = 9.99          # 修改副本中第 0 行的净值
print(table)                 # 原表的第 0 行也变了
print(table[0] is table2[0]) # 两个表的第 0 行是同一个对象
```

需要连内层一起复制时，用标准库 `copy` 模块中的 `deepcopy()`，它会沿着引用把每一层都复制一份。本课程用到嵌套列表的场合不多，这里了解即可；第 3 周 pandas 数据表的 `copy()` 默认就是深复制。

```python
import copy
table3 = copy.deepcopy(table)
table3[0][1] = 1.0
print(table)                 # 原表不受影响
print(table3)
```

运算符 `is` 判断两个名字是否指向同一个对象（即编号是否相同），`==` 判断两个对象的内容是否相等。两个内容相同的列表可以是不同的对象。

```python
a = [1, 2]
b = a
c = [1, 2]
print(a is b, a == b)
print(a is c, a == c)
```

这一区别在处理数据表时会反复出现：写 `df2 = df` 之后对 `df2` 删列、填补缺失值，原来的 `df` 也随之改变。第 3 周会再谈这个问题。

### 2.7 判断元素是否存在

运算符 `in` 判断某个值是否是列表的元素，结果是布尔值。方法 `count()` 返回某个值出现的次数，`index()` 返回它第一次出现的索引。

```python
codes = ["000001", "000003", "000001"]
print("000003" in codes)
print("000002" in codes)
print(codes.count("000001"))
print(codes.index("000003"))
```

### 2.8 字符串也是序列

字符串可以看作字符的序列，因此列表的索引、切片、`len()` 和 `in` 同样适用于字符串。

```python
s = "华夏成长混合"
print(s[0])
print(s[-2:])
print(len(s))
print("成长" in s)
```

但字符串是不可变的，不能按位置修改。下面这一格会报错。

```python
s[0] = "中"
```

`TypeError: 'str' object does not support item assignment`：字符串对象不支持按位置赋值。要“修改”字符串，只能生成一个新的字符串，例如用 `replace()` 方法；原字符串不受影响。

```python
t = s.replace("混合", "股票")
print(t)
print(s)
```

方法 `split()` 按指定的分隔符把字符串切成列表，`join()` 反过来把列表中的字符串用指定的连接符拼成一个字符串。数据文件中常见的 CSV 格式，每一行就是用逗号分隔的若干字段。

```python
line = "000001,华夏成长,1.2345"
fields = line.split(",")
print(fields)
print(len(fields))
print(" | ".join(fields))
```

列表一节到此结束。它比后面三种容器的篇幅长得多，原因是索引、切片、可变与不可变、原地修改这些机制都在这里第一次出现；后面的元组、字典、集合只需说明它们与列表的不同之处。

列表可以修改，这既是便利也是风险：一个列表可能在程序的某处被无意改动。有些数据本来就不应被修改，例如一条完整的记录。Python 为此提供了不可变的序列——元组。

## 3 元组

元组（tuple）与列表类似，元素按位置排列，可以索引和切片；区别在于元组一旦建立就不能修改。元组写在圆括号内。与列表一样，元组对元素的类型没有限制，可以混合，也可以嵌套。

```python
record = ("000001", "2024-01-31", 1.2345)
print(record[0])
print(record[-1])
print(len(record))
print(type(record))
```

下面这一格会报错。

```python
record[2] = 1.3
```

`TypeError: 'tuple' object does not support item assignment`，与字符串的情形相同。

元组常用于表示一条记录，或者把几个值打包在一起传递（第 7 节的函数会用它一次返回多个值）。把元组的元素一次赋给几个变量，称为解包（unpacking）；变量的个数必须与元素个数一致。

```python
code, date, nav = record
print(code)
print(date)
print(nav)
```

只有一个元素的元组要在元素后加逗号，否则圆括号只被当作普通的括号。

```python
print(type((1)))
print(type((1,)))
```

列表和元组都按位置查找元素。但许多数据天然是“名称对应取值”的形式：每个基金代码对应一个基金名称，每个类别对应一个数量。若把它们放进列表，查找某个代码的名称就需要先知道它排在第几位，很不方便。字典解决的就是这个问题。

## 4 字典

字典（dict）由若干“键：值”对组成，写在花括号内。通过键（key）查找对应的值（value），而不是通过位置。键在字典中必须唯一。

```python
names = {"000001": "华夏成长", "000003": "中海可转债"}
print(names["000001"])
print(len(names))
print(type(names))
```

用方括号查找一个不存在的键会报错。下面这一格会报错。

```python
names["000002"]
```

`KeyError: '000002'`，表示字典中没有这个键。如果不确定键是否存在，可以用 `get()` 方法：键不存在时返回 `None`，或者返回指定的默认值，而不报错。

```python
print(names.get("000002"))
print(names.get("000002", "未知"))
```

给字典中不存在的键赋值，就是新增一对键值；给已存在的键赋值，则覆盖原来的值。`del` 语句删除一对键值。`in` 运算符判断的是键是否存在，不是值。

```python
names["000002"] = "新基金"          # 新键：新增
names["000001"] = "华夏成长混合"    # 已有的键：覆盖
print(names)
```

```python
print("000002" in names)
print("华夏成长混合" in names)     # in 只查键，不查值
del names["000002"]
print(len(names))
```

`keys()`、`values()`、`items()` 三个方法分别给出字典的全部键、全部值和全部键值对。这里用 `list()` 把它们转成列表打印出来；对它们逐个处理的写法在第 6 节介绍。

```python
print(list(names.keys()))
print(list(names.values()))
print(list(names.items()))
```

字典对键和值的类型要求不同。值没有任何限制：可以是数字、字符串、列表，甚至另一个字典。因此一条包含多个字段的记录也常用字典表示，字段名作键。

```python
fund = {"code": "000001", "name": "华夏成长", "navs": [1.00, 1.05, 1.02]}
print(fund["name"])
print(fund["navs"])
print(fund["navs"][-1])      # 先按键取出列表，再按位置取最后一个
```

键则必须是不可变类型。字符串、数字、元组都可以作键；列表不能。原因在于字典是靠键的值来定位数据的，如果键在放入之后还能被修改，字典就找不到它了。元组作键在数据处理中很常见，例如用“代码与月份”的组合唯一标识一条记录。

```python
monthly = {}
monthly[("000001", "2024-01")] = 1.05
monthly[("000001", "2024-02")] = 1.02
print(monthly)
print(monthly[("000001", "2024-02")])
```

用列表作键会报错。下面这一格会报错。

```python
monthly[["000001", "2024-03"]] = 1.08
```

`TypeError: unhashable type: 'list'`。“unhashable”（不可哈希）是 Python 对“不能作字典键”的说法；所有可变类型都是不可哈希的。

有时我们不关心每个元素对应什么值，只关心哪些元素出现过：一批交易记录里涉及了哪些基金，两张表的基金代码有多少是共同的。这类问题用集合处理最方便。

## 5 集合

集合（set）中的元素互不重复，且没有顺序。把一个列表转换成集合，重复的元素会自动合并；因此比较列表的长度与其集合的长度，可以判断列表中是否有重复。

```python
codes = ["000001", "000002", "000001", "000003", "000002"]
uniq = set(codes)
print(uniq)
print(len(codes), len(uniq))
print(len(codes) == len(set(codes)))
```

集合支持数学上的交、并、差运算，分别写作 `&`、`|`、`-`。比较两批代码时，交集是双方共有的，差集是一方独有的。

```python
a = {"000001", "000002"}
b = {"000002", "000003"}
print(a & b)
print(a | b)
print(a - b)
```

集合没有顺序，打印出来的先后不代表任何含义，也不能用索引访问。需要按确定顺序处理时，先用 `sorted()` 转成有序列表。

```python
print(sorted(uniq))
print("000001" in uniq)
```

集合的元素与字典的键受同样的限制：必须是不可变类型。数字、字符串、元组可以放进集合，列表不可以，原因相同——集合靠元素的值判断它是否已经存在。

### 四种容器的比较

| 容器 | 写法 | 构造函数 | 查找方式 | 有序 | 允许重复 | 可修改 | 元素类型 |
| --- | --- | --- | --- | --- | --- | --- | --- |
| 列表 list | `[1, 2, 3]` | `list()` | 按位置 | 是 | 是 | 是 | 任意，可混合、可嵌套 |
| 元组 tuple | `(1, 2, 3)` | `tuple()` | 按位置 | 是 | 是 | 否 | 任意，可混合、可嵌套 |
| 字典 dict | `{"a": 1}` | `dict()` | 按键 | 保持插入顺序 | 键不重复 | 是 | 键须不可变；值任意 |
| 集合 set | `{1, 2, 3}` | `set()` | 只判断是否存在 | 否 | 否 | 是 | 须不可变 |

选用哪种容器取决于数据的性质和要做的操作：按位置取用一串同类数据，用列表；一条不应修改的记录，用元组；按名称查找，用字典；只关心成员是否存在，用集合。

表中“构造函数”一列值得单独说明。`list`、`tuple`、`dict`、`set` 既是类型的名字，也可以像函数一样调用，作用是把给它的东西变成这种类型。前面已经用过几次：`set(codes)` 把列表变成集合以去重，`list(names.keys())` 把字典的键变成列表。转换的对象可以是任何能逐个取出元素的东西，包括字符串。

```python
print(list("abc"))
print(tuple([1, 2, 3]))
print(set([1, 1, 2, 3, 3]))
print(dict([("000001", "华夏成长"), ("000003", "中海可转债")]))
```

不给参数时，构造函数建立一个空容器：`list()` 与 `[]` 相同，`dict()` 与 `{}` 相同。空集合只能写 `set()`——因为 `{}` 已经被空字典占用了。

```python
print(type([]), type(list()))
print(type({}))              # 空的花括号是字典，不是集合
print(type(set()))
```

第一讲用过的 `int()`、`float()`、`str()` 是同一回事：它们也是类型名，调用它们就是把值转换成该类型。类型名可调用、调用即转换，是 Python 的一条通则。

有了容器之后，接下来的问题是如何对容器中的每个元素做同样的处理，以及如何根据条件决定执行哪些操作。这就是控制流。

## 6 控制流

到目前为止，代码都是从上到下逐行执行，每行执行一次。控制流（control flow）语句改变这一点：条件语句让某些代码只在条件成立时执行；循环语句让一段代码重复执行多次。

### 6.1 条件语句

`if` 语句的形式是：`if` 后面写一个条件（结果为布尔值的表达式），以冒号结尾；下面缩进的若干行是条件为真时执行的语句块。可以用 `elif`（else if 的缩写）接续其他条件，用 `else` 处理所有条件都不成立的情形。

在 Python 中，缩进是语法的一部分，用来表示语句块的范围。同一语句块中的各行必须缩进相同的空格数，惯例是 4 个空格；JupyterLab 在冒号后回车时会自动缩进。

```python
r = -0.02
if r > 0:
    print("上涨")
elif r == 0:
    print("持平")
else:
    print("下跌")
```

条件语句的一个常见用途是在做某个运算之前先检查它是否可行。例如计算收益率要除以上一期净值，若上一期净值为 0，除法会报错；数据中的 0 往往表示缺失而不是真正的 0，应当先判断。

```python
nav_prev = 0.0
nav_now = 1.05
if nav_prev > 0:
    print(nav_now / nav_prev - 1)
else:
    print("上期净值为 0，视为缺失")
```

### 6.2 for 循环

`for` 循环对容器中的每个元素依次执行同一段代码。写法是 `for 变量 in 容器:`，下面缩进的语句块是循环体。每一轮，Python 从容器中取出下一个元素赋给循环变量，然后执行循环体；元素取完，循环结束。

```python
for nav in [1.00, 1.05, 1.02]:
    print(nav * 2)
```

这个循环执行了三轮：第一轮 `nav` 是 1.00，打印 2.0；第二轮 `nav` 是 1.05，打印 2.1；第三轮 `nav` 是 1.02，打印 2.04。循环结束后 `nav` 仍保留最后一轮的值。

若需要按次数循环或按索引访问，用 `range()`。`range(n)` 依次产生 0 到 n−1；`range(a, b)` 产生 a 到 b−1，同切片一样含头不含尾；`range(a, b, step)` 指定步长。`range` 产生的不是列表，需要查看全部内容时可以用 `list()` 转换。

```python
for i in range(4):
    print(i)
```

```python
print(list(range(1, 4)))
print(list(range(0, 10, 3)))
```

循环时经常既需要元素，又需要它的索引。`enumerate()` 同时给出两者。

```python
navs = [1.00, 1.05, 1.02, 1.10]
for i, nav in enumerate(navs):
    print(i, nav)
```

`zip()` 把两个（或多个）序列按位置配对，每轮同时取出各序列的对应元素。

```python
codes = ["000001", "000003"]
fund_names = ["华夏成长", "中海可转债"]
for c, n in zip(codes, fund_names):
    print(c, n)
```

对字典循环，取出的是键；用 `items()` 可以同时取出键和值。

```python
names = {"000001": "华夏成长", "000003": "中海可转债"}
for code in names:
    print(code, names[code])
```

```python
for code, name in names.items():
    print(code, "->", name)
```

### 6.3 用循环计算收益率

净值序列 `navs` 中，第 t 期的简单收益率定义为 `navs[t] / navs[t-1] - 1`。第 0 期没有上一期，所以收益率从 t = 1 开始计算；n 个净值只能得到 n − 1 个收益率。用循环实现时，循环变量 `i` 从 1 取到 `len(navs) - 1`，即 `range(1, len(navs))`。

```python
navs = [1.00, 1.05, 1.02, 1.10]
rets = []
for i in range(1, len(navs)):
    r = navs[i] / navs[i - 1] - 1
    rets.append(r)
print(len(navs), len(rets))
print(rets)
```

逐轮追踪：`i = 1` 时计算 1.05 / 1.00 − 1 = 0.05；`i = 2` 时计算 1.02 / 1.05 − 1 ≈ −0.0286；`i = 3` 时计算 1.10 / 1.02 − 1 ≈ 0.0784。`rets` 最终有 3 个元素。结果的小数位很长，可以用 `round()` 保留四位（下面用到的写法是列表推导式，见 6.5 节）。

```python
print([round(r, 4) for r in rets])
```

如果误把循环范围写成 `range(len(navs))`，`i = 0` 那一轮会计算 `navs[0] / navs[-1]`——负索引 `-1` 指向最后一个元素，于是得到一个并不存在的“第 0 期收益率”，而且结果有 4 个而不是 3 个。这是收益率计算中最常见的错误。检查结果的长度是否为 n − 1，是发现它的最快办法。

```python
wrong = []
for i in range(len(navs)):
    wrong.append(navs[i] / navs[i - 1] - 1)
print(len(wrong))
print([round(r, 4) for r in wrong])
```

同一计算也可以用 2.3 节的两个切片配合 `zip()` 完成：`navs[:-1]` 是各期的前一期净值，`navs[1:]` 是各期的后一期净值，按位置配对后逐对相除。

```python
rets2 = []
for prev, curr in zip(navs[:-1], navs[1:]):
    rets2.append(curr / prev - 1)
print([round(r, 4) for r in rets2])
```

### 6.4 累加与累乘

对一组数求和用内置函数 `sum()`，求最大、最小值用 `max()`、`min()`。但多期收益率的累计不是求和：期初 1 元，经过收益率 r₁、r₂、…之后变为 (1 + r₁)(1 + r₂)…，累计收益率是这个乘积减 1。涨 10% 再跌 10%，结果不是回到原点，而是亏损 1%。

```python
rets = [0.1, -0.1]
print(sum(rets))
total = 1.0
for r in rets:
    total = total * (1 + r)
print(round(total - 1, 4))
```

循环体中的 `total = total * (1 + r)` 是一种累积写法：每轮把上一轮的结果乘上一个因子再存回同一个变量。累加的写法与之类似：`total = total + x`，也可以简写为 `total += x`。

### 6.5 列表推导式

“对容器中的每个元素计算一个新值，收集成新列表”这种循环极为常见，Python 为它提供了简洁的写法——列表推导式（list comprehension）：`[表达式 for 变量 in 容器]`。下面两种写法等价。

```python
squares = []
for x in range(5):
    squares.append(x ** 2)
print(squares)

squares2 = [x ** 2 for x in range(5)]
print(squares2)
```

推导式末尾可以加 `if` 条件，只保留满足条件的元素。

```python
rets = [0.02, -0.01, 0.03, -0.04]
positive = [r for r in rets if r > 0]
print(positive)
```

用推导式重写 6.3 节的收益率计算只需一行。

```python
navs = [1.00, 1.05, 1.02, 1.10]
rets = [curr / prev - 1 for prev, curr in zip(navs[:-1], navs[1:])]
print([round(r, 4) for r in rets])
```

### 6.6 计数：字典与循环的配合

统计一组类别数据中每个类别出现的次数，是数据处理中最常见的任务之一。用字典保存“类别：次数”，循环遍历数据：每遇到一个类别，就把它对应的次数加 1；第一次遇到某个类别时字典里还没有这个键，用 `get(t, 0)` 取默认值 0。

```python
types = ["股票型", "混合型", "股票型", "债券型", "股票型"]
counts = {}
for t in types:
    counts[t] = counts.get(t, 0) + 1
print(counts)
```

逐轮追踪：第一轮 `t` 是“股票型”，字典为空，`get` 返回 0，加 1 后存入；第二轮“混合型”同理；第三轮又是“股票型”，`get` 返回 1，加 1 得 2；依此类推，循环结束时得到各类别的计数。第 3 周会看到 pandas 用一个方法完成这件事，它内部所做的与这里相同。

### 6.7 while 循环

`for` 循环适用于遍历已知的容器或已知的次数。当循环次数事先未知、只知道终止条件时，用 `while` 循环：只要条件为真，就一直重复执行循环体。例如，年化收益 8%，问多少年后资产翻倍。

```python
nav = 1.0
years = 0
while nav < 2.0:
    nav = nav * 1.08
    years = years + 1
print(years)
print(round(nav, 3))
```

写 `while` 循环时必须确保条件最终会变为假，否则循环永不结束。若在 JupyterLab 中不慎陷入无限循环，用工具栏的停止按钮中断内核。在数据处理中，`while` 的使用远少于 `for`。

前面计算收益率的代码已经写了三遍。当同一段逻辑在程序中反复出现时，应当把它定义为函数，此后只需调用函数名，而不必重复代码。

## 7 函数

### 7.1 定义与调用

函数（function）是一段有名字的、可以反复调用的代码。之前用过的 `len()`、`print()`、`type()`、`round()`、`sorted()` 都是 Python 内置的函数。用 `def` 语句可以定义自己的函数：`def` 后面是函数名和圆括号中的参数列表，以冒号结尾；缩进的语句块是函数体；`return` 语句指定函数的返回值。

```python
def simple_return(p0, p1):
    return p1 / p0 - 1
```

执行这一格没有任何输出：`def` 只是定义了函数，并没有运行它。调用函数时，在函数名后的括号里给出实际的值。调用的过程是：把这些值依次赋给参数 `p0`、`p1`，执行函数体，遇到 `return` 就把后面表达式的值作为整个调用表达式的值返回。

```python
print(simple_return(1.00, 1.25))
x = simple_return(4, 6)
print(x)
```

第一次调用中 `p0` 是 1.00、`p1` 是 1.25，返回 0.25；第二次调用返回 0.5 并赋给了 `x`。函数体中的 `p0`、`p1` 只在函数内部有意义，每次调用重新赋值。

### 7.2 返回值与打印

初学者常混淆 `return` 与 `print`。`print` 把值显示在屏幕上，仅此而已——对程序而言，它没有产生任何可以继续使用的值。`return` 则把值交回调用处，可以赋给变量、参与后续计算。一个函数如果没有 `return` 语句，调用它得到的值是 `None`。

```python
def return_by_print(p0, p1):
    print(p1 / p0 - 1)

y = return_by_print(4, 6)
print(y)
```

第一行输出的 0.5 来自函数内部的 `print`；第二行输出的 `None` 才是 `y` 的值。此时若想继续用 `y` 计算，会报错。下面这一格会报错。

```python
y + 1
```

`TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`：`None` 不能与整数相加。屏幕上看到了正确的数字，程序中却没有得到这个数字——这就是把 `print` 当作 `return` 的后果。

### 7.3 返回多个值

`return` 后面写多个用逗号分隔的值，函数就返回一个元组；调用处可以用第 3 节的解包一次接收。

```python
def min_max(xs):
    return min(xs), max(xs)

result = min_max([1.0, 1.05, 0.98])
print(result)
print(type(result))

lo, hi = min_max([1.0, 1.05, 0.98])
print(lo, hi)
```

### 7.4 参数的默认值与关键字参数

定义函数时可以给参数指定默认值。调用时若不提供该参数，就使用默认值；提供了则用提供的值。调用时还可以写明参数名（称为关键字参数），这样不必记住参数的顺序，代码也更易读。

```python
def annualize(r, periods=12):
    return (1 + r) ** periods - 1

print(round(annualize(0.01), 4))              # periods 取默认值 12：月度收益年化
print(round(annualize(0.01, 252), 4))         # periods 为 252：日度收益年化
print(round(annualize(0.01, periods=4), 4))   # 关键字参数：季度收益年化
```

有默认值的参数必须写在没有默认值的参数之后。默认值实际上是函数编写者替调用者做出的假定：月度还是日度、含不含费用。使用别人写的函数时，应当查看每个参数的默认值是否符合自己的情形。

### 7.5 函数内部对参数的修改

2.6 节说过，赋值不复制对象。把一个列表传给函数时也是如此：函数的参数只是指向同一个列表的另一个名字。如果函数体修改了这个列表，调用处的列表也就被修改了。

```python
def apply_fee(prices, fee=0.01):
    for i in range(len(prices)):
        prices[i] = prices[i] * (1 - fee)
    return prices

p = [100.0, 200.0]
q = apply_fee(p)
print(p)
print(q is p)
```

调用之后 `p` 本身变成了扣费后的价格，`q` 与 `p` 是同一个对象。函数除返回值之外对外部造成的改变称为副作用（side effect）。如果调用者之后还需要原始的 `p`，这个函数就造成了错误。若不希望修改传入的数据，应在函数开头复制一份，对副本操作。

```python
def apply_fee_safe(prices, fee=0.01):
    prices = list(prices)              # 复制；此后 prices 指向新列表
    for i in range(len(prices)):
        prices[i] = prices[i] * (1 - fee)
    return prices

p = [100.0, 200.0]
q = apply_fee_safe(p)
print(p)
print(q)
```

### 7.6 作用域

函数内部定义的变量（包括参数）只在函数内部有效，称为局部变量；函数外部的变量称为全局变量。函数内部可以读取全局变量的值。

```python
total = 100

def show():
    print(total)

show()
```

但如果在函数内部对某个名字赋值，Python 就把它当作局部变量——即使外部有同名的全局变量。下面的函数试图累加全局变量 `total`，会报错。

```python
total = 0

def add(x):
    total = total + x
    return total

add(5)
```

`UnboundLocalError: cannot access local variable 'total' where it is not associated with a value`。由于函数体中出现了 `total = ...`，`total` 被认定为局部变量；而在计算右边的 `total + x` 时，这个局部变量还没有被赋过值。

正确的做法不是在函数内部修改全局变量，而是让函数返回计算结果，由调用处决定如何使用。

```python
total = 0

def add(x):
    return total + x

total = add(5)
print(total)
```

这一约定使函数成为一个独立的单元：它的行为只取决于输入，结果只通过返回值传出。这样的函数容易理解、容易检查，也容易被其他代码复用。

有了循环和函数，已经能够处理列表中的数据。但数据分析要面对的往往是几万、几百万个数。对这样的数据逐个元素循环，代码冗长，运行也慢。NumPy 提供了另一种做法：把整列数据当作一个对象，一次完成整列的运算。

## 8 NumPy 数组

### 8.1 为什么需要数组

先看列表的局限。要把列表中的每个数都乘以 2，不能直接写 `navs * 2`——对列表而言 `*` 表示重复。

```python
navs = [1.00, 1.05, 1.02, 1.10]
print(navs * 2)
```

必须写成循环或推导式。

```python
print([x * 2 for x in navs])
```

NumPy 是 Python 中数值计算的基础库，Anaconda 已经装好。它提供了一种新的容器——数组（ndarray）。数组上的算术运算是逐元素进行的：`arr * 2` 就是把每个元素乘以 2。使用 NumPy 前先导入它；`import numpy as np` 把这个库以简称 `np` 引入，此后用 `np.` 访问其中的函数。这是通行的约定。

### 8.2 dtype 与 shape

数组与列表的一个根本区别是：数组中的所有元素必须是同一类型。这个类型称为数组的 dtype。正因为元素类型一致、在内存中连续存放，NumPy 才能用底层的高效代码一次处理整列数据。用混有整数和小数的列表建立数组时，全部元素会被统一为浮点数。

```python
print(navs.dtype)
print(np.array([1, 2, 3]).dtype)
print(np.array([1, 2.5]).dtype)
print(np.array(["000001", "000003"]).dtype)   # <U6 表示长度不超过 6 的字符串
```

数组的形状（shape）是一个元组，给出每个维度的长度。一维数组的形状形如 `(4,)`。

```python
print(navs.shape)
print(navs.ndim)
print(len(navs))
```

除了从列表转换，还可以直接生成数组：`np.arange()` 类似 `range()`，`np.zeros()` 和 `np.ones()` 生成全 0、全 1 数组，`np.linspace()` 在区间内生成等距的点。

```python
print(np.arange(5))
print(np.zeros(3))
print(np.ones(3))
print(np.linspace(0, 1, 5))
```

### 8.3 逐元素运算

数组与一个数做算术，运算作用于每个元素；两个形状相同的数组做算术，对应位置的元素两两运算。

```python
a = np.array([1.0, 2.0, 3.0])
b = np.array([10.0, 20.0, 30.0])
print(a + 4)
print(a * b)
print(b / a)
```

对比列表：`+` 对列表是拼接，对数组是逐元素相加。同一个符号，含义由操作对象的类型决定，这一点在 1.1 节已经说过；读代码时必须先弄清操作对象是列表还是数组。

```python
print([1, 2, 3] + [4])
print(np.array([1, 2, 3]) + 4)
```

### 8.4 索引、切片与收益率

数组的索引和切片规则与列表完全相同。6.3 节用循环计算收益率，用数组只需一行：后 n−1 个元素除以前 n−1 个元素，再减 1。

```python
r = navs[1:] / navs[:-1] - 1
print(r)
print(r.shape)
```

结果有 3 个元素，与 4 个净值对应。这里没有写循环，但计算的内容与 6.3 节完全一样：`navs[1:]` 是各期的后一期净值，`navs[:-1]` 是前一期净值，两个数组按位置逐元素相除。NumPy 还提供 `np.diff()` 计算相邻元素之差。

```python
print(np.diff(navs))
print(np.diff(navs) / navs[:-1])
```

### 8.5 汇总运算

数组有 `sum()`、`mean()`、`std()`、`min()`、`max()` 等方法，对全部元素汇总。`round()` 方法按指定小数位数对每个元素四舍五入。

```python
print(r.sum())
print(r.mean())
print(r.min(), r.max())
print(r.round(4))
print(r.mean().round(4), r.std().round(4))
```

### 8.6 布尔数组与筛选

比较运算对数组逐元素进行，结果是一个由 `True`、`False` 组成的布尔数组。它有三种典型用法。第一，对布尔数组求和，得到满足条件的元素个数（1.3 节说过 `True` 按 1 计）；第二，求均值，得到满足条件的比例；第三，把布尔数组放进方括号中作为索引，只取出对应位置为 `True` 的元素——这就是数据筛选。

```python
r = np.array([0.02, -0.01, 0.03, -0.04])
print(r > 0)
print((r > 0).sum())
print((r > 0).mean())
print(r[r > 0])
print(r[r > 0].mean())
```

多个条件组合时，对数组不能用 `and`、`or`，要用 `&`（且）、`|`（或），并且每个条件必须加括号，因为 `&` 的优先级高于比较运算符。

```python
print(r[(r > -0.02) & (r < 0.025)])
```

### 8.7 二维数组与 axis

数组可以是二维的，相当于一个矩阵或一张表。在金融数据中，习惯上每一行是一个时间点、每一列是一个资产。下面的数组记录 3 个月、2 只基金的月收益率。二维数组的形状是 `(行数, 列数)`；`m[0]` 取第 0 行，`m[:, 0]` 取第 0 列（逗号前是对行的选择，逗号后是对列的选择，冒号表示全取）。

```python
m = np.array([[0.01, 0.02],
              [0.03, -0.02],
              [0.02, 0.06]])
print(m.shape)
print(m[0])
print(m[:, 0])
print(m[1, 1])
```

对二维数组汇总时，需要说明沿哪个方向汇总，用参数 `axis` 指定。`axis=0` 表示沿着行的方向（即对每一列的所有行）汇总，结果的长度等于列数；`axis=1` 表示沿着列的方向（对每一行的所有列）汇总，结果的长度等于行数；不指定 `axis` 则对全部元素汇总。可以这样记：`axis` 等于几，就是把第几个维度“压掉”。

```python
print(m.mean(axis=0))   # 每一列（每只基金）在 3 个月上的平均：长度为 2
print(m.mean(axis=1))   # 每一行（每个月）在 2 只基金上的平均：长度为 3
print(m.mean())
```

`axis` 写反是常见错误。检查的办法是看结果的长度：求每只基金的平均，结果的元素个数应等于基金数。

### 8.8 缺失值 NaN

数据中的缺失值在 NumPy 中用 `np.nan`（Not a Number）表示。NaN 有两个需要注意的性质。第一，它参与任何算术运算的结果都是 NaN，因此含有 NaN 的数组求 `mean()` 得到 NaN；要忽略缺失值计算，用 `np.nanmean()`、`np.nansum()` 等函数。第二，NaN 与任何值都不相等，包括它自身，因此不能用 `== np.nan` 判断缺失，要用 `np.isnan()`。

```python
x = np.array([1.0, np.nan, 3.0])
print(x.mean())
print(np.nanmean(x))
print(np.nan == np.nan)
print(np.isnan(x))
print(np.isnan(x).sum())     # 缺失值的个数
```

### 8.9 累乘

6.4 节用循环计算累计收益。`np.cumprod()` 返回累积乘积：第 k 个元素是前 k 个元素的乘积。对 `1 + r` 累乘，就得到各期期末的累计净值；减 1 即累计收益率。

```python
r = np.array([0.1, -0.1])
print(np.cumprod(1 + r))
print(np.cumprod(1 + r) - 1)
```

### 8.10 除以零：数组运算不会中断

用循环计算收益率时，若某期净值为 0，Python 会抛出 `ZeroDivisionError` 并停止。数组运算则不同：NumPy 只给出一行警告（`RuntimeWarning`，不是错误），把结果记为 `inf`（无穷大），程序继续执行。这意味着错误的数据不会让程序停下，而是一直传递到最终结果中。

```python
nav = np.array([1.0, 1.1, 0.0, 1.2, 1.26])
r = nav[1:] / nav[:-1] - 1
print(r)
print(np.isinf(r).any())     # 结果中是否有 inf
```

结果中出现了 −1（分子为 0）和 `inf`（分母为 0），两者都不是真实的收益率。正确的做法是在计算前把表示缺失的 0 替换为 NaN。`np.where(条件, 值1, 值2)` 逐元素判断：条件为真取值 1，否则取值 2，返回新数组。替换后，涉及缺失值的两期收益率都成为 NaN，其余照常。

```python
nav = np.where(nav > 0, nav, np.nan)
r = nav[1:] / nav[:-1] - 1
print(r)
print(np.nanmean(r))
```

本节介绍的数组是第 3 周 pandas 的基础：pandas 的 Series 可以理解为带有标签的一维数组，DataFrame 是共用行标签的若干列。dtype、布尔筛选、axis、NaN 这些概念在 pandas 中原样出现。

```python
# M3
def winsorize(rs, cap=0.1):
    for i in range(len(rs)):
        if rs[i] > cap:
            rs[i] = cap
    return rs
orig = [0.05, 0.3, -0.2]
w = winsorize(orig)
print(orig, w)
```
